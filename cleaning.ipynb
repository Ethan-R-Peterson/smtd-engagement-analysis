**Cleaning**

```python
# Drop unnecessary columns
df = df.drop(columns=[
    "amount attributed",
    "Imported Department",
    "Imported Grad Year",
    "Imported Minor",
    "Registrar Minor",
    "Status",
    "Registrar Grad Code",
    "Registrar Admit Code",
    "Registrar Program"
])

# Rename columns for clarity
df = df.rename(columns={
    "Experience_fk": "experience_id",
    "Imported Class Year": "imported_class_year",
    "Imported Program": "imported_program",
    "Imported Unit": "imported_unit",
    "Major Dept Incl PAME": "major_department",
    "Major Unit": "major_unit",
    "Notes": "notes",
    "Person-fk": "person_id",
    "Record ID": "record_id",
    "Registrar Class Year": "registrar_class_year",
    "Registrar Department": "registrar_department",
    "Registrar First Name": "registrar_first_name",
    "Registrar Last Name": "registrar_last_name",
    "Registrar Unit": "registrar_unit",
    "Role": "role",
    "unique participant": "unique_participant",
    "Experiences:: Name": "experience_name",
    "Experiences:: Term": "experience_term",
    "Experiences:: Type": "experience_type",
    "Experiences:: Year": "experience_year",
    "Participant Experience distinct::event term": "participant_experience_term_unique",
    "Participant Experience distinct::event type": "participant_experience_type_unique",
    "Participant Experience distinct::event year": "participant_experience_year_unique",
    "Participant Experience distinct::Major Dept Incl PAME": "participant_major_department_unique",
    "Participant Experience distinct::Major Unit": "participant_major_unit_unique",
    "Participant Experience distinct::Programs": "participant_programs_unique",
    "Participant Experience distinct::RecordID": "participant_record_id_unique",
    "People::Uniqname": "uniqname",
    "Registrar Data::Department": "registrar_department",
    "Registrar Data::Program": "registrar_program",
    "Registrar Data::Type": "registrar_type",
    "Registrar Data::Unit": "registrar_unit"
})

df.head()
```

```python
# -- Clean major_department (keep only the first major if multiple are listed) --
acceptable_majors = {
    "Winds & Percussion",
    "Theatre & Drama",
    "Strings",
    "Composition",
    "Voice",
    "Piano",
    "Dance",
    "Performing Arts Technology",
    "Musical Theatre",
    "Jazz",
    "Conducting",
    "Music, Theatre & Dance",
    "Ross School of Business",
    "Chamber Music",
    "LSA",
    "Music Education",
    "Computer Science",
    "Information",
    "Stamps School of Art & Design",
    "Economics",
    "Communication and Media",
    "Psychology",
    "Musicology",
    "Film, Television, and Media",
    "Biopsych, Cognit & Neurosci",
    "Mechanical Engineering",
    "Political Science",
    "Music",
    "Entrepreneurship & Leadership",
    "Education",
    "Organ"
}

def extract_first_major(entry):
    if pd.isna(entry):
        return entry
    sorted_majors = sorted(acceptable_majors, key=len, reverse=True)
    remaining = entry
    while remaining:
        for major in sorted_majors:
            if remaining.startswith(major):
                return major
        # No match found at start, skip one character and try again
        remaining = remaining[1:]
    return entry  # leave unchanged if no match found

df["major_department"] = df["major_department"].apply(extract_first_major)
```

```python
# Attempt forward filling missing values of different columns
df = df.ffill()
df.isna().sum().sort_values(ascending=False) # check if any nan values, make sure column names correct

df.head(20)
```

```python

# -- Clean term --
# Separate season and year from experience_term
df = df.dropna(subset=["experience_term"])
df["term_season"] = df["experience_term"].str[:2]
df["year"] = df["experience_term"].str[-4:].astype(int)

# Combine student first + last name
df["student_name"] = df["registrar_first_name"] + " " + df["registrar_last_name"]
```

```python
# --- Clean class year ---
def clean_class_year(x):
    x = str(x).strip()

    remove = ["Pro 3rd Yr", "USpec/NCFD", "GNCFD/MIGS", "Pro 2nd Yr", "School Year", "Grad Cert"]
    if x in remove:
        return None
    
    grad_terms = ["G Pre-Cand", "G Cand", "G Post-Cand", "Grad Inter", "Doctorate", "Masters", "Grad Mastr"]
    if x in grad_terms:
        return "Graduate"

    valid = ["Freshman", "Sophomore", "Junior", "Senior"]
    if x in valid:
        return x

    return None

df["class_year"] = df["imported_class_year"].apply(clean_class_year)

# Drop unwanted rows
df = df.dropna(subset=["class_year"])
```

```python
# Drop more unnecessary columns
df = df.drop(columns=[
    "imported_class_year",
    "registrar_department",
    "registrar_unit",
    "imported_unit",
    "imported_program",
])
```
