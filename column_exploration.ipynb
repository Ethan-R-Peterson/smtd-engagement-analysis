```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("Filemaker_CLEAN.csv")
```

```python
pd.set_option('display.max_rows', None)

counts = df["experience_type"].value_counts()
print(len(counts)) 
print(counts)         
```

```python
counts = df["experience_name"].value_counts()

grouped = df.groupby("experience_name")["experience_type"].unique()

# reorder by most popular experience_name
result = grouped.loc[counts.index]

print(result)
```

```python
# major_department: duplicate names, many majors. 
pd.set_option('display.max_rows', None)
counts = df["major_department"].value_counts()
filtered = counts[counts > 50]

print(len(filtered))
print(filtered)
```
